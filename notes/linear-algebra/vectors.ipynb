```python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])
print(a+b)
print(a-b)
print(3*a)
```

```python
a = np.array([3, 4])
b = np.array([1, 2])
print(np.dot(a, b))
print(np.linalg.norm(a))
print(np.dot(a, a))
print(np.linalg.norm(a) ** 2)
```
