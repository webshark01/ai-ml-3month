```python
import numpy as np

A = np.array([[2, 0],
              [0, 3]])

eigenvalues, eigenvectors = np.linalg.eig(A)
print(eigenvalues)
print(eigenvectors)
```

```python
A = np.array([[4, 0],
              [0, 7]])
eigenvalues, eigenvectors = np.linalg.eig(A)
v = eigenvectors[:,0]
iam = eigenvalues[0]

print(A @ v)
print(iam * v)
```
